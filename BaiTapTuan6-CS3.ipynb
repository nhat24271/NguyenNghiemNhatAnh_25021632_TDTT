```python
#W6A9
dict_1=input().split()
dict_2=input().split()
all_dict=dict_1 + dict_2
new_dict={}
for pair in all_dict:
    k, v = pair.split(':')
    v=int(v)
    if k in new_dict:
        new_dict[k]+=v
    else:
        new_dict[k]=v
sort_new_dict=dict(sorted(new_dict.items()))
print(sort_new_dict)
```

```python
#W6A10
mylist=list(map(int, input().split()))
k=int(input())
while k<=0:
    print('vui long nhap so k nguyen duong')
    k=int(input())
kq=[]
s=len(mylist)
for i in range(s):
    for x in range(i+1,s):
        if mylist[i] + mylist[x]==k:
            pair=(mylist[i], mylist[x])
            if pair not in kq:
                kq.append(pair)
kq.sort()
print(kq)
```

```python
#w6A13
def rsvdict(input_list):
    new_dict={}   
    for pair in input_list:
        k, v = pair.split(':')
        new_dict[v]=k
    return new_dict 

mydict=input().split()
kq=rsvdict(mydict)
print(kq)
```

```python
#W6A14
def giao(a, b):
    kq=[]
    for nums in b:
        if nums in a:
            kq.append(nums)
    return kq
    
list_A = [int(x) for x in input().split()]
list_B = [int(y) for y in input().split()]
ds=giao(list_A, list_B)
print(ds)
```

```python
#W6A15
def ktra(a:dict, b:int):
    m={}
    for k, v in a.items():
        if b<v:
            m[k]=v
    return m     
    
mydict= {k: int(v) for k, v in (pair.split(':') for pair in input().split())}
k=int(input())
kq=ktra(mydict, k)
print(kq)    
```
