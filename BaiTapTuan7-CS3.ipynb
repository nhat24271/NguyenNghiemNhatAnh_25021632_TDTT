```python
#W7A1
arr=[int(x) for x in input().split()]
left=0
right=len(arr)-1
target=int(input())
def binary_search(arr, left, right, target):
    if right>=left:
        mid=left+(right-left)//2
        if arr[mid]==target:
            return mid
        elif arr[mid]>target:
            return binary_search(arr, left, mid-1, target)
        elif arr[mid]<target:
            return binary_search(arr, mid+1, right, target)
print(binary_search(arr, left, right, target))
```

```python
#W7A2
def count_occurrences(arr , x):
    cnt=0
    for nums in arr:
        if nums ==x:
            cnt+=1
    return cnt
arr=[int(n) for n in input().split()]
x=int(input())
print(count_occurrences(arr, x))
```

```python
#W7A3
def bubble_sort(arr):
    for i in range(len(arr)):
        for j in range(len(arr)-i-1):
            if arr[j]>arr[j+1]:
                arr[j], arr[j+1]=arr[j+1], arr[j]
    return arr
arr=list(map(int, input().split()))
print(bubble_sort(arr))
```

```python
#W7A3
def selection_sort(arr):
    for i in range(len(arr)):
        min_idx=i
        for j in range(i+1,len(arr)):
            if arr[j]<arr[min_idx]:
                min_idx=j
        arr[i], arr[min_idx]=arr[min_idx], arr[i]
    return arr
arr=list(map(int, input().split()))
print(selection_sort(arr))
```

```python
#W7A4
def search(arr):
    m={}
    for num in arr:
        if num in m:
            m[num]+=1
        else:
            m[num]=1
    max_val=max(m.values())
    for k, v in m.items():
        if v== max_val:
            return f'{k} xuat hien nhieu nhat, som nhat, {v} lan'
arr=[int(x) for x in input().split()]
print(search(arr))
```

```python
#W7A5
arr=[int(x) for x in input().split()]
n=int(input())
def w7a5(arr, n):
    m={}
    for num in arr:
        if num in m:
            m[num]+=1
        else:
            m[num]=1
    cnt=0
    for k, v in m.items():
        needed=n-k
        if k<needed:
            cnt+=v*m[needed]
        elif k==needed:
            cnt+=v*(v-1)/2
    return cnt
print(w7a5(arr, n))
```

```python
#W7A7
arr=[int(x) for x in input().split()]
def w7a7(arr):
    n=len(arr)
    m=[1]*n
    for i in range(n):
        for j in range(i):
            if arr[i]>arr[j]:
                m[i]=max(m[i],m[j]+1)
    return max(m) if arr else 0
print(w7a7(arr))
```

```python
#W7A8
interval=eval(input())
queries=[int(x) for x in input().split()]
def w7a8(interval:list, queries:list):
    m=[-1]*len(queries)
    for i,sublist in enumerate(interval):
        for j in range(len(queries)):
            if queries[j] in range(sublist[0],sublist[1]+1):
                d=sublist[1]-sublist[0]+1
                if m[j]==-1:
                    m[j]=d
                else:
                    m[j]=min(m[j],d)
    return m
print(w7a8(interval,queries))
```

```python
#W7A9
nums=[int(x) for x in input().split()]
def w7a9(nums):
    n=len(nums)
    m=[0]*n
    for i in range(n):
        cnt=0
        for j in range(i+1,n):
            if nums[j]<nums[i]:
                cnt+=1
        m[i]=cnt
    return m
print(w7a9(nums))
```

```python
#W7A10
def w7a10(arr, m):
    n=len(arr)
    m1=[]
    for i in range(n):
        for j in range(i,n):
            du=sum(arr[i:j])%m
            m1.append(du)
    return max(m1)
arr=[int(x) for x in input().split()]
m=int(input())
print(w7a10(arr, m))
```

```python
#W7A13
n=int(input())
arr=[int(x) for x in input().split()]
def w7a13(arr):
    myDict={}
    for num in arr:
        if num in myDict:
            myDict[num]+=1
        else:
            myDict[num]=1
    tempMax=0
    for num in arr:
        if num+1 in myDict:
            temp=myDict[num]+myDict[num+1]
            if temp>tempMax:
                tempMax=temp
    return tempMax
print(w7a13(arr))
```

```python
#W7A14
n=int(input())
arr=[int(x) for x in input().split()]
def w7a14(arr):
    m=[]
    for i, val in enumerate(arr):
        if val==7:
            m.append(i)
    if not m:
        return "Not found"
    else:
        m1=sorted(m,reverse=True)
        return ' '.join(map(str, m1))
print(w7a14(arr))
```

```python
#W7A15
n=int(input())
myList=[]
for _ in range(n):
    myList.append(input())
def nemo(myList):
    nemoIdx=myList.index('Nemo')
    result=[]
    result.append(myList[nemoIdx-1])
    if nemoIdx==len(myList)-1:
        result.append(myList[0])
    else:
        result.append(myList[nemoIdx+1])
    return " and ".join(result)
print(nemo(myList))
```
